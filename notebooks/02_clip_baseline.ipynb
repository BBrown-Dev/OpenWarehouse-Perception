# 02 - Zero-Shot SigLIP Baseline

Image-level AUROC by prompt condition. `scripts/run_poc.py` produces the scores; this notebook loads them. Conditions were selected on dev before held-out summaries could be computed.

```python
import pandas as pd
from IPython.display import Image, display
RESULTS = "../docs/results/"
FIGURES = "../docs/figures/"
pd.set_option("display.width", 200, "display.max_columns", 30)
```

```python
dev = pd.read_csv(RESULTS + "siglip_dev.csv")
dev.pivot(index="class_name", columns="condition", values="auroc").round(3)
```

```python
held = pd.read_csv(RESULTS + "siglip_heldout.csv")
held.pivot(index="class_name", columns="condition", values="auroc").round(3)
```

```python
held.groupby("condition")["auroc"].mean().round(3).sort_values(ascending=False)
```

`delta_low` and `delta_high` are the paired CI against the `article` baseline. The frozen `photo` condition does not beat it.
